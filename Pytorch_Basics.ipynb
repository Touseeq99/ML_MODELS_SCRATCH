```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
from sklearn.datasets import load_diabetes
```

```python
df = load_diabetes()
```

```python
df
```

```python
df1 = pd.DataFrame(df.data , columns = df.feature_names)
```

```python
df1['Target'] = df.target
```

```python
df1.head(2)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score
```

```python
df1.info()
```

```python
x = df1.drop('Target' , axis=1)
y = df1['Target']
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=28)
```

```python
x_train.shape
```

```python
y_train.shape
y_test.shape
```

```python
scaler = StandardScaler()
```

```python
x_scale_train=scaler.fit_transform(x_train)
x_scale_test=scaler.transform(x_test)
```

```python
import torch.nn as nn
import torch.optim as op
```

```python
x_tensor = torch.tensor(x_scale_train , dtype=torch.float32)
x_test_tensor = torch.tensor(x_scale_test , dtype=torch.float32)
y_tensor = torch.tensor(y_train.values.reshape(-1, 1) , dtype=torch.float32)
y_test_tensor = torch.tensor(y_test.values.reshape(-1, 1) , dtype=torch.float32)
```

```python
print(y_test_tensor)
avg = y_test_tensor.mean()
print(avg)
```

### **MODEL LAYER**

```python
layer1=nn.Linear(x_scale_train.shape[1],1)
```

### ***LOSS LAYER***

```python
loss_fn = nn.MSELoss()
```

```python
optimizer = op.Adam(layer1.parameters() , lr=0.01)
```

```python
n_epochs = 5000
```

```python
for i in range(n_epochs):
  pred = layer1(x_test_tensor)
  loss = loss_fn(pred , y_test_tensor)
  print(loss)
  optimizer.zero_grad()
  loss.backward()
  optimizer.step()
```

```python
from sklearn.metrics import r2_score , mean_squared_error
```

```python
pred = layer1(x_test_tensor)
```

```python
pred1 = np.array(pred.detach().numpy())
result1 = r2_score(y_test_tensor , pred1)
result2 = mean_squared_error(y_test_tensor , pred1)
```

```python
print('-----MSE-----' , result2)
print('-----R2-----' , result1)
```
